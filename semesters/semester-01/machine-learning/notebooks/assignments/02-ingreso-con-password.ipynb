# Tarea 02 · Ingreso a un sitio mediante password

**Semestre 1 · Módulo 01 · Machine Learning**

## Enunciado

> Por favor construya un ingreso a un sitio, debe decir si logró o no ingresar al sitio para eso usted debe tener un Password sencillo. Mientras no tenga el password debe mostrar un mensaje diciendo, no es el password correcto; hasta que no tenga el password correcto no lo deja ingresar al sitio.

## Objetivo

Practicar el bucle `while` con un **contador de intentos**, la decisión con `if`/`else`, la
interrupción del bucle con `break` y una condición final para el caso de intentos agotados.

## Solución

La siguiente celda contiene la solución original entregada, sin modificaciones. El password es
`1234` y se permiten hasta 3 intentos.

```python
password = "1234"

intentos = 0

while intentos < 3:

    clave = input("Digite el password númerico: ")

    if clave == password:
        print("Logró ingresar al sitio")
        break
    else:
        print("No es el password correcto")
        intentos = intentos + 1

if intentos == 3:
    print("No logró ingresar al sitio, por favor comuniquese con el administrador")
```

## Cómo se ejecuta

- `while intentos < 3:` repite la petición mientras queden intentos.
- `if clave == password:` compara lo digitado con el password; si coincide, imprime el mensaje
  de éxito y `break` sale del bucle de inmediato.
- `else:` cuando no coincide, avisa del error y suma 1 a `intentos`.
- `if intentos == 3:` al terminar el bucle sin acierto, muestra el mensaje para contactar al
  administrador.
