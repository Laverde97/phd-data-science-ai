IMPORTANDO LIBRERÍAS

```python
import numpy as np;
import matplotlib.pyplot as plt;
import pandas as pd;
```

```python
dataset = pd.read_csv('https://cdtiueb.com.co/doctorado/datos_icfes.csv');
print(dataset);
X = dataset.iloc[:, :-1].values;
y = dataset.iloc[:, 7].values;
```

SE MUESTRA EL CONTENIDO DE LA VARIABLE X

```python
X
```

```python
y
```

Preparando los datos, Imputar datos

```python
import sklearn;
imputer = sklearn.impute.SimpleImputer(missing_values = np.nan, strategy = 'mean');
imputer = imputer.fit(X[:, 2:7]);
X[:, 2:7] = imputer.transform(X[:, 2:7]);

```

```python
X
```

CREAR DATAFRAME CON DATOS LIMPIOS

```python
df_limpio = pd.DataFrame(X, columns=['documento', 'departamento', 'matematicas', 'sociales', 'ciencias', 'lectura', 'ingles'])
df_limpio['aprobado'] = y

print("--- DATOS CON VALORES FALTANTES IMPUTADOS ---")
print(df_limpio)
print("\n")
```

```python
# Aplicar get_dummies
df_codificado = pd.get_dummies(df_limpio, columns=['departamento', 'aprobado'], drop_first=True)
print(df_codificado);
# Convertir columnas booleanas a int (1 y 0)
for col in df_codificado.columns:
    if df_codificado[col].dtype == 'bool':
        df_codificado[col] = df_codificado[col].astype(int)

print("--- DATOS CODIFICADOS (SIN REORDENAR) ---")
print(df_codificado)
print("\n")
```

REORDENAR COLUMNAS: DUMMY PRIMERO, aprobado_SI ÚLTIMA

VERIFICAR ESTRUCTURA FINAL

```python
# Identificar la columna documento
columna_documento = ['documento']
# Identificar columnas dummy de departamento (las que empiezan con 'departamento_')
columnas_departamentos = [col for col in df_codificado.columns if col.startswith('departamento_')]
# Identificar la columna aprobado_SI
columna_aprobado = ['aprobado_SI']
# Identificar columnas numéricas ('matematicas', 'sociales', 'ciencias', 'lectura', 'ingles')
columnas_numericas = ['matematicas', 'sociales', 'ciencias', 'lectura', 'ingles']

# Crear nuevo orden: primero documento, luego las columnas dummy de departamento, luego las numéricas, y compro_SI al final
#nuevo_orden = columna_documento + columnas_departamentos + columnas_numericas + columna_aprobado
nuevo_orden = columna_documento + columnas_departamentos + columnas_numericas +columna_aprobado
#print(nuevo_orden);
# Reordenar el DataFrame
df_codificado = df_codificado[nuevo_orden]

print("--- DATOS CODIFICADOS (aprobado_SI como última columna) ---")
print(df_codificado)
print("\n")
```

```python
print("--- ESTRUCTURA FINAL DEL DATAFRAME ---")
print(f"🔹 Total de filas: {df_codificado.shape[0]}")
print(f"🔹 Total de columnas: {df_codificado.shape[1]}")
print(f"🔹 Columnas en orden: {list(df_codificado.columns)}")
print("\n")
```

```python
df_codificado.to_csv('DATOS_ICFES_CODIFICADO.csv', index=False)
print("✅ Archivo guardado como 'DATOS_ICFES_CODIFICADO.csv'")
```
