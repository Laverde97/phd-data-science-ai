```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

importando el conjunto de datos

```python
dataset_ICFES=pd.read_json('https://cdtiueb.com.co/doctorado/datos_icfes.json')
(dataset_ICFES)
```

```python
dataset_ICFES.info()
```

Datos faltantes

```python
dataset_ICFES.isnull().sum()
```

```python
dataset_ICFES = dataset_ICFES.replace(r'^\s*$', np.nan, regex=True)
```

```python
dataset_ICFES.isnull().sum()
```

```python
X=dataset_ICFES.iloc[:,:-1].values # En X en mayuscula ya que es el set de los datos
print("============================================== \n data en X")
print(X)
```

```python
y=dataset_ICFES.iloc[:,-1].values
print("============================================== \n data en y")
print(y)
```

```python
print("============================================== \n data en X")
print(X)

```

```python
print("============================================== \n data en y")
print(y)
```

```python
import sklearn
imputer=sklearn.impute.SimpleImputer(missing_values=np.nan,strategy='mean')
imputer=imputer.fit(X[:,2:7])
X[:,2:7]=imputer.transform(X[:,2:7])
print(X)
```

```python
X
```

## CREAR DATAFRAME CON DATOS LIMPIOS

```python
df_limpio=pd.DataFrame(X,columns=['Documento','Departamento','Matematicas','Sociales','Ciencias','Lectura','Ingles'])
df_limpio['Aprobo']=y

print("--- DATOS CON VALORES FALTANTES INPUTADOS ---")
print(df_limpio)
print('\n')
```

##CODIFICAR VARIABLES CATEGORICAS CON pd.get_dummies

```python
# Aplicar get_dummies
df_codificado=pd.get_dummies(df_limpio, columns=['Departamento', 'Aprobo'], drop_first=True)
print("--- DATOS CODIFICADOS ---")
print(df_codificado)
print('\n')

# Convertir columnas booleanas a int (1 y 0)
for col in df_codificado.columns:
    if df_codificado[col].dtype == 'bool':
        df_codificado[col] = df_codificado[col].astype(int)

print("--- DATOS CODIFICADOS (SIN REORDENAR) ---")
print(df_codificado)
print('\n')
```

## REORDENAR COLUMNAS: DUMMY primero, COMPRO_SI ÚLTIMA

```python
#Identificar columnas dummy de Depto_ (Las que empiezan con 'Depto_')
columnas_Depto=[Col for Col in df_codificado.columns if Col.startswith('Departamento_')]
print(columnas_Depto)
# Identificar la columna Aprobo_SI
Columna_Aprobo= ['Aprobo_SI']

# Identificar columnas numéricas (Edad y Salario)
Columna_Numericas= ['Matematicas','Sociales','Ciencias','Lectura','Ingles']
ID_DOCUMENTO= ['Documento']
# Crear nuevo orden: Primero las dummy de pais, luego las numericas y compro?SI al final
Nuevo_Orden= ID_DOCUMENTO + columnas_Depto + Columna_Numericas + Columna_Aprobo

# Reordenar el DataFrame
df_reordenado=df_codificado[Nuevo_Orden]

print("--- DATOS CODIFICADOS (REORDENADOS) ---")
print(df_reordenado)
print('\n')
```

## VERIFICAR ESTRUTURA FINAL


```python
print("--- ESTRUCTURA FINAL DEL DATAFRAME ---")
print(f"🔹 Total de filas: {df_codificado.shape[0]}")
print(f"🔹 Total de columnas: {df_codificado.shape[1]}")
print(f"🔹 Columnas en orden: {list(df_codificado.columns)}")
print("\n")
```
