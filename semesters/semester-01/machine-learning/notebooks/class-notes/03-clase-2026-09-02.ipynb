##MATPLOTLIB


### ¿Qué es Matplotlib?

**Matplotlib** es una biblioteca completa para crear visualizaciones estáticas, animadas e interactivas en Python. Es ampliamente utilizada para la visualización de datos, ofreciendo un conjunto de herramientas flexible y extenso para generar varios tipos de gráficos, como gráficos de líneas, gráficos de dispersión, gráficos de barras, histogramas, gráficos 3D y mucho más.

### ¿Para qué sirve Matplotlib?

*   **Versatilidad:** Admite una gran variedad de tipos de gráficos, lo que la hace adecuada para casi cualquier necesidad de visualización.
*   **Salida de alta calidad:** Puede generar figuras con calidad de publicación en varios formatos impresos y entornos interactivos.
*   **Personalización:** Casi todos los elementos de un gráfico (por ejemplo, colores, estilos de línea, fuentes, etiquetas de ejes, leyendas) se pueden controlar, lo que permite visualizaciones altamente personalizadas.
*   **Integración:** Se integra bien con otras bibliotecas de Python, especialmente NumPy para operaciones numéricas y Pandas para la manipulación de datos.

```python
import matplotlib.pyplot as plt
import numpy as np

# Datos para el gráfico
x = np.linspace(0, 2 * np.pi, 100) # 100 puntos de 0 a 2*pi
y = np.sin(x)

# Crear una figura y un eje
plt.figure(figsize=(8, 4))

# Trazar los datos
plt.plot(x, y, label='sin(x)', color='blue', linestyle='--')

# Añadir etiquetas y título
plt.xlabel('Eje X')
plt.ylabel('Eje Y')
plt.title('Gráfico simple de onda sinusoidal con Matplotlib')

# Añadir una leyenda
plt.legend()

# Añadir una cuadrícula
plt.grid(True)

# Mostrar el gráfico
plt.show()
```

```python
import matplotlib.pyplot as plt
x=[3,6,7,8];
y=[6,14,8,19];
plt.plot(x,y, color='red', linewidth=3, label='Ejemplo línea');
plt.legend();
plt.grid();
plt.xlabel('Eje X');
plt.ylabel('Eje Y');
plt.title('algo en X vs algo en Y');
plt.show();
```

```python
import matplotlib.pyplot as plt
x=[3,6,7,8];
y=[6,14,8,19];
plt.bar(x,y, color='red', linewidth=3, label='Ejemplo línea');
plt.legend();
plt.grid();
plt.xlabel('Eje X');
plt.ylabel('Eje Y');
plt.title('algo en X vs algo en Y');
plt.show();
```

```python
import matplotlib.pyplot as plt
x=[3,6,7,8];
y=[6,14,8,19];
plt.scatter(x,y, label='Ejemplo línea');
plt.legend();
plt.grid();
plt.xlabel('Eje X');
plt.ylabel('Eje Y');
plt.title('algo en X vs algo en Y');
plt.show();
```

```python
import matplotlib.pyplot as plt
#primera linea
x=[1,2,3,4]
y1=[6,14,8,19]
plt.plot(x,y1, color='blue', linewidth=3, label='costos')

#segunda linea
y2=[9,15,15,24]
plt.plot(x,y2, color='red', linewidth=3, label='ventas')

#Ahora vamos a definir el título y el nombre de los ejes
plt.title('Ejemplo diagrama de lineas')
plt.xlabel('Eje x')
plt.ylabel('Eje y')

#Mostrar el gráfico y la grid
plt.legend()
plt.grid()
plt.show()
```
