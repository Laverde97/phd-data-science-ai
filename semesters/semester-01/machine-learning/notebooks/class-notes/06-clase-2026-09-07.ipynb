## MACHINE LEARNING (PREPARANDO LOS DATOS)


## Importando las librerias

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

importando el conjunto de datos

```python
dataset=pd.read_csv('https://cdtiueb.com.co/doctorado/DATOS1.txt')
#dataset= pd.read_csv('DATOS1.csv');
print(dataset)
```

```python
X=dataset.iloc[:,:-1].values # En X en mayuscula ya que es el set de los datos
print("============================================== \n data en X")
print(X)
y=dataset.iloc[:,-1].values
print("============================================== \n data en y")
print(y)
```

```python
X
```

```python
y
```

PREPARANDO DATOS, IMPUTAR DATOS

```python
import sklearn
imputer=sklearn.impute.SimpleImputer(missing_values=np.nan,strategy='mean')
imputer=imputer.fit(X[:,1:3])
X[:,1:3]=imputer.transform(X[:,1:3])
print(X)
```

```python
X
```

## CREAR DATAFRAME CON DATOS LIMPIOS

```python
df_limpio=pd.DataFrame(X,columns=['Pais','Edad','Salario'])
df_limpio['Compro']=y

print("--- DATOS CON VALORES FALTANTES INPUTADOS ---")
print(df_limpio)
print('\n')
```

##CODIFICAR VARIABLES CATEGORICAS CON pd.get_dummies

```python
# Aplicar get_dummies
df_codificado=pd.get_dummies(df_limpio, columns=['Pais', 'Compro'], drop_first=True)
print("--- DATOS CODIFICADOS ---")
print(df_codificado)
print('\n')

# Convertir columnas booleanas a int (1 y 0)
for col in df_codificado.columns:
    if df_codificado[col].dtype == 'bool':
        df_codificado[col] = df_codificado[col].astype(int)

print("--- DATOS CODIFICADOS (SIN REORDENAR) ---")
print(df_codificado)
print('\n')
```

## REORDENAR COLUMNAS: DUMMY primero, COMPRO_SI ÚLTIMA

```python
#Identificar columnas dummy de pais (Las que empiezan con 'Pais_')
columnas_Pais=[Col for Col in df_codificado.columns if Col.startswith('Pais_')]
print(columnas_Pais)

# Identificar la columna Compro_SI
Columna_Compro= ['Compro_SI']

# Identificar columnas numéricas (Edad y Salario)
Columna_Numericas= ['Edad', 'Salario']

# Crear nuevo orden: Primero las dummy de pais, luego las numericas y compro?SI al final
Nuevo_Orden= columnas_Pais + Columna_Numericas + Columna_Compro

# Reordenar el DataFrame
df_reordenado=df_codificado[Nuevo_Orden]

print("--- DATOS CODIFICADOS (REORDENADOS) ---")
print(df_reordenado)
print('\n')
```

## VERIFICAR ESTRUTURA FINAL


```python
print("--- ESTRUCTURA FINAL DEL DATAFRAME ---")
print(f"🔹 Total de filas: {df_codificado.shape[0]}")
print(f"🔹 Total de columnas: {df_codificado.shape[1]}")
print(f"🔹 Columnas en orden: {list(df_codificado.columns)}")
print("\n")
```

## GUARDAR RESULTADOS

```python
df_codificado.to_csv('DATOS1_CODIFICADO.csv', index=False)
print("✅ Archivo guardado como 'DATOS1_CODIFICADO.csv'")
```
