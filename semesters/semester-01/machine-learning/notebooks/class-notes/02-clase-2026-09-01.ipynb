```python
import numpy as np
A= np.array([-1,2,7,4,5])
print(A)
```

Un **`np.array`** (o arreglo de NumPy) es la estructura de datos fundamental en la librería NumPy. Es similar a una lista de Python, pero con algunas diferencias clave:

*   **Homogeneidad**: Todos los elementos en un `np.array` deben ser del mismo tipo de datos (por ejemplo, todos enteros, todos flotantes, etc.). Esto permite a NumPy optimizar las operaciones.
*   **Eficiencia**: Los arreglos de NumPy están implementados en C, lo que los hace mucho más eficientes en términos de rendimiento y uso de memoria para operaciones numéricas, especialmente con grandes volúmenes de datos, en comparación con las listas de Python.
*   **Funcionalidades matemáticas**: `np.array` viene con una gran cantidad de funciones matemáticas y operaciones vectorizadas que se pueden aplicar a todo el arreglo de manera eficiente, sin necesidad de bucles explícitos en Python.

**¿Para qué se usa?**

Se utiliza ampliamente en:
*   **Cálculo numérico**: Para realizar operaciones matemáticas complejas de forma rápida y eficiente.
*   **Análisis de datos**: Es la base de muchas otras librerías como Pandas, Scikit-learn, etc.
*   **Machine Learning**: Almacenar y manipular características de datos, pesos de modelos, etc.
*   **Procesamiento de imágenes y señales**: Representar datos multidimensionales.

En el código `A = np.array([-1, 2, 7, 4, 5])`, estás creando un arreglo de NumPy unidimensional con los números enteros -1, 2, 7, 4 y 5.

## LISTA

Una **lista en Python** es una colección ordenada y mutable de elementos. Esto significa lo siguiente:

*   **Ordenada**: Los elementos tienen un orden definido, y ese orden no cambia a menos que se modifique explícitamente la lista. Puedes acceder a los elementos por su índice (posición).
*   **Mutable**: Puedes cambiar, añadir o eliminar elementos de la lista después de que ha sido creada. No es una estructura fija.
*   **Colección**: Puede contener cualquier tipo de datos de Python, incluso mezclados (enteros, flotantes, cadenas de texto, otras listas, etc.).

**¿Para qué sirve una lista?**

Las listas son extremadamente útiles y se utilizan para:

1.  **Almacenar colecciones de datos**: Cuando necesitas guardar múltiples ítems que están relacionados entre sí, como una lista de nombres de estudiantes (`lista1 = ['Maria', 'Sofia', 'Laura', 'Carmen']`), una serie de números, o una colección de objetos.
2.  **Iteración**: Son perfectas para recorrer sus elementos usando bucles `for`, como en el ejercicio que planteaste para mostrar todo el contenido de la lista.
3.  **Manipulación de datos**: Permiten realizar operaciones como añadir elementos (`append()`, `insert()`), eliminar (`remove()`, `pop()`), ordenar (`sort()`), invertir (`reverse()`), buscar (`index()`), y más.
4.  **Implementar otras estructuras de datos**: Se pueden usar como base para implementar pilas, colas, o matrices simples antes de recurrir a librerías más especializadas como NumPy.

```python
lista1= ['Maria', 'Sofia', 'Laura', 'Carmen'];
print(lista1[0]);
print(lista1[1]);
print(lista1[2]);
print(lista1[3]);
```

## AGREGAR ELEMENTOS A LA LISTA

```python
lista1= ['Maria', 'Sofia', 'Laura', 'Carmen','Pedro'];
lista1.append('Juan');
lista1.append('Carlos');

for i in range(len(lista1)): # Corregido: Usar range(len(lista1)) para iterar sobre los índices válidos
  print(lista1[i]);
```

```python
listae = []
print("Por favor, ingresa 5 elementos para la lista:")
for i in range(5):
  elemento = input(f"Elemento {i+1}: ")
  listae.append(elemento)

print("\nElementos actuales de la lista:")
print(listae)

nombre_a_buscar = input("\nIngresa un nombre para buscar en la lista: ")

if nombre_a_buscar in listae:
  print(f"'{nombre_a_buscar}' se encontró en la lista.")
  indice = listae.index(nombre_a_buscar)
  listae.pop(indice)
  print("Lista después de eliminar el elemento:")
  print(listae)
else:
  print(f"'{nombre_a_buscar}' no se encontró en la lista.")
  print("La lista permanece sin cambios:")
  print(listae)
```
