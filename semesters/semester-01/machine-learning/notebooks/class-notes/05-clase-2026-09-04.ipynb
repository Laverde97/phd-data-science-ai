Imprimiendo los valores

La función `iloc` en Pandas se utiliza para la indexación basada en enteros puros. Permite seleccionar filas y columnas por su posición entera, es decir, por su índice numérico de 0 a N-1.

### ¿Qué significa `iloc`?

`iloc` significa "**integer location**" (ubicación por entero).

### ¿Cómo se usa?

La sintaxis básica es `df.iloc[filas, columnas]`:

*   **`filas`**: Puede ser un solo entero, una lista de enteros, un objeto de `slice` (como `1:5`), o un valor booleano.
*   **`columnas`**: Puede ser un solo entero, una lista de enteros, un objeto de `slice` (como `1:5`), o un valor booleano.

Si omites las columnas (es decir, `df.iloc[filas]`), por defecto seleccionará todas las columnas para las filas especificadas.

### Ejemplos en el notebook:

*   `df.iloc[0,:]` selecciona la primera fila (índice 0) y todas las columnas.
*   `df.iloc[0,1]` selecciona el elemento en la primera fila (índice 0) y la segunda columna (índice 1).
*   `df.iloc[:,:]` selecciona todas las filas y todas las columnas, es decir, el DataFrame completo.
*   `df.iloc[:2,:]` selecciona las primeras dos filas (índices 0 y 1) y todas las columnas.
*   `df.iloc[:3,:2]` selecciona las primeras tres filas (índices 0, 1 y 2) y las primeras dos columnas (índices 0 y 1).
*   `df.iloc[1:,:]` selecciona desde la segunda fila (índice 1) hasta el final y todas las columnas.
*   `df.iloc[2:,:]` selecciona desde la tercera fila (índice 2) hasta el final y todas las columnas.
*   `df.iloc[3:,:]` selecciona la cuarta fila (índice 3) hasta el final y todas las columnas.
*   `df.iloc[1:,2:]` selecciona desde la segunda fila (índice 1) hasta el final y desde la tercera columna (índice 2) hasta el final.
*   `df.iloc[:,1:]` selecciona todas las filas y desde la segunda columna (índice 1) hasta el final.
*   `df.iloc[:,1]` selecciona todas las filas de la segunda columna (índice 1).

REALIZAR UN ANALISIS CON ESTA BASE DE DATOS


La utilidad generada por cada cliente y generar diferentes gráficas.

https://storage.googleapis.com/bucket_gv_20260902_ueb/datos18.xlsx

```python
# ============================================================
# ANÁLISIS DE LA UTILIDAD GENERADA POR CLIENTE
# ============================================================

# 1. Importación de librerías
import pandas as pd
import matplotlib.pyplot as plt


# 2. Carga de la base de datos
url = "https://storage.googleapis.com/bucket_gv_20260902_ueb/datos18.xlsx"

Clientes = pd.read_excel(url)


# 3. Exploración inicial de la información
print("Dimensiones de la base de datos")
print("-" * 35)
print("Número de filas:", Clientes.shape[0])
print("Número de columnas:", Clientes.shape[1])

print("\nPrimeros registros:")
display(Clientes.head())

print("\nVariables disponibles:")
print(Clientes.columns.tolist())
```

Cálculo de la utilidad

La utilidad se obtiene como la diferencia entre el valor de la factura y el costo de la factura:

```python
# CALCULA DE LA UTILIDAD

# 5. Cálculo de la utilidad
Clientes["Utilidad"] = (
    Clientes["valor_factura"] - Clientes["costo_factura"]
)

Clientes.head()
```

```python
# GRÁFICA DE UTILIDAD VS. CLIENTES

plt.figure(figsize=(12, 6))

plt.bar(Clientes["nombre_cliente"], Clientes["Utilidad"])

plt.title("Utilidad generada por cliente", fontsize=14)
plt.xlabel("Cliente", fontsize=12)
plt.ylabel("Utilidad", fontsize=12)

plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

## Costo de factura por cliente

```python
# GRÁFICA: COSTO DE FACTURA POR CLIENTE

plt.bar(Clientes["nombre_cliente"], Clientes["costo_factura"])

plt.title("Costo de factura por cliente", fontsize=14)
plt.xlabel("Cliente")
plt.ylabel("Costo de factura")

plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

## Valor de factura por cliente

```python
# GRÁFICA: VALOR DE FACTURA POR CLIENTE

plt.figure(figsize=(12, 6))

plt.bar(Clientes["nombre_cliente"], Clientes["valor_factura"])

plt.title("Valor de factura por cliente", fontsize=14)
plt.xlabel("Cliente")
plt.ylabel("Valor de factura")

plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

## Comparación entre costo y valor de factura por cliente

```python
# GRÁFICA: COSTO VS VALOR DE FACTURA POR CLIENTE

Clientes.plot(
    x="nombre_cliente",
    y=["costo_factura", "valor_factura"],
    kind="bar",
    figsize=(14, 6)
)

plt.title("Comparación entre costo y valor de factura por cliente", fontsize=14)
plt.xlabel("Cliente")
plt.ylabel("Valor")

plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

## Relación entre costo y valor de factura

```python
# GRÁFICA: RELACIÓN ENTRE COSTO Y VALOR DE FACTURA

plt.figure(figsize=(9, 6))

plt.scatter(
    Clientes["costo_factura"],
    Clientes["valor_factura"]
)

plt.title("Relación entre costo y valor de factura", fontsize=14)
plt.xlabel("Costo de factura")
plt.ylabel("Valor de factura")

plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# GRÁFICA GENERAL: COSTO, VALOR DE FACTURA Y UTILIDAD POR CLIENTE

Clientes.plot(
    x="nombre_cliente",
    y=["costo_factura", "valor_factura", "Utilidad"],
    kind="bar",
    figsize=(14, 7)
)

plt.title(
    "Comparación de costo, valor de factura y utilidad por cliente",
    fontsize=14
)

plt.xlabel("Cliente", fontsize=12)
plt.ylabel("Valor", fontsize=12)

plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)

plt.legend([
    "Costo de factura",
    "Valor de factura",
    "Utilidad"
])

plt.tight_layout()
plt.show()
```

```python
# GRÁFICO DE DISPERSIÓN:
# COSTO DE FACTURA VS VALOR DE FACTURA

plt.figure(figsize=(10, 6))

plt.scatter(
    Clientes["costo_factura"],
    Clientes["valor_factura"],
    s=80,
    alpha=0.7
)

# Agregar el nombre de cada cliente
for i in range(len(Clientes)):
    plt.annotate(
        Clientes["nombre_cliente"].iloc[i],
        (
            Clientes["costo_factura"].iloc[i],
            Clientes["valor_factura"].iloc[i]
        ),
        fontsize=8
    )

plt.title(
    "Relación entre costo y valor de factura por cliente",
    fontsize=14
)

plt.xlabel("Costo de factura", fontsize=12)
plt.ylabel("Valor de factura", fontsize=12)

plt.grid(alpha=0.3)
plt.tight_layout()

plt.show()
```
