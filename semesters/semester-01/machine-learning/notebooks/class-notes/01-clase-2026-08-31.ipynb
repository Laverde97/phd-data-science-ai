Guardar en las variables a, b c, cual es el mayor y cual es el menor, utilizando el input para realizarlo en python


```python
a = float(input("Ingrese el valor de a: "))
b = float(input("Ingrese el valor de b: "))
c = float(input("Ingrese el valor de c: "))

# Encontrar el mayor
if a >= b and a >= c:
    mayor = a
elif b >= a and b >= c:
    mayor = b
else:
    mayor = c

# Encontrar el menor
if a <= b and a <= c:
    menor = a
elif b <= a and b <= c:
    menor = b
else:
    menor = c

print("El número mayor es:", mayor)
print("El número menor es:", menor)
```

Realizar un programa que me sume los primeros n números pares

---



```python
n = int(input("Ingrese la cantidad de números pares a sumar (n): "))

suma_pares = 0
contador = 0
numero_actual = 2 # Empezamos con el primer número par

while contador < n:
    suma_pares = suma_pares + numero_actual
    numero_actual = numero_actual + 2 # Siguiente número par
    contador = contador + 1

print(f"La suma de los primeros {n} números pares es: {suma_pares}")
```

CONSTRUIR UN PROGRAMA PARA QUE ME HAGA LA TABLA DE MULTIPLICAR CON DIGITAR EL N, EL QUE UNO QUIERA HACER, HASTA DONDE QUIERE LLEGAR, POR EJEMPLO, N=5 HASTA 20


```python
n = int(input('Digite la tabla que quiere realizar: ')) #Esta línea le pide al usuario el número de la tabla de multiplicar.
hasta = int(input('Digite hasta dónde quiere llegar: ')) # Aquí se pregunta hasta qué número queremos multiplicar.

x = 1 # La variable x es nuestro contador, comienza en 1 porque queremos empezar multiplicando: 5 x 1

while x <= hasta: # Mientras x sea menor o igual que hasta, repita las instrucciones que están dentro del while.
    print(n, 'x', x, '=', n * x) # "x": Es el simbolo, x: Hasta donde quiere multiplicar, "=": Es el simbolo, n * x: Es la multiplicación de 5 * 3, que da 15
    x = x + 1 # Esta línea aumenta el contador en 1, si no se pone, x siempre sería 1 y el programa repetiría para siempre: 5 x 1 = 5
```
